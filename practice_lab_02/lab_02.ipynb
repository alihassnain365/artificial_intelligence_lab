Write a function to find mean, median, mode for the given set of numbers in a list.
1. Read the elements into a list.
2. Calculate the sum of list elements.
3. Calculate the mean, median.
4. Display the result.

```python
import statistics
def stat_calc(numbers:list):
    """Shos the mean, median and mode of the given list"""
    print(f"Mean : {statistics.mean(numbers)}")
    print(f"Median : {statistics.median(numbers)}")
    print(f"Mode : {statistics.mode(numbers)}")
transactions = [100,200,300,500,12,45,785,293,82]
stat_calc(transactions)
```

```python
Write a function dups to find all duplicates in the list.
1. Create a list to read elements.
2. Pass the list as parameter to dup function.
3. Define function dup to identify duplicate elements in the list.
4. Return the final list to called function.
   
```

```python
def dup(numbers:list):
    """Returns the duplicate numbers in the list"""
    unique_list = set(numbers)
    duplicate = list()
    seen = set()
    for num in numbers:
        if num in seen:
            duplicate.append(num)
        else:
            seen.add(num)
    return duplicate

transactions = [12,12, 23,2029,39,284,29,304,29,294,283,249,38,49,38,45,22,4,5,555,32,5983,245,2,67,8,98]
print(f"Duplicates are : {dup(transactions)}")
        
    
```
